```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("demo").getOrCreate()
```

```python
#PySpark join-related question
# Sample Data

customers_data = [("John", "CA"), ("Alice", "NY"), ("Bob", "TX"), ("Masx", "CA")]
orders_data = [("John",250 ),("Alice",300 ), ("Bob",400 ), ("Kay",300 ) ]
```

```python
# Create Data Frame

customrs_df = spark.createDataFrame(data = customers_data, schema=["name", "state"])
orders_df = spark.createDataFrame(data = orders_data, schema = ["name", "amount"])

```

```python
# Now see the dataset

customrs_df.show()
orders_df.show()
```

```python
# Now Joins the data

inner_join_df = customrs_df.join(orders_df, on = "name", how="inner")
inner_join_df.show()
```

```python
# Outer JOin

outer_df = customrs_df.join(orders_df, on = "name", how="outer")
outer_df.show()
```

```python
# Left Join

left_join_df = customrs_df.join(orders_df, on = "name", how = "left")
left_join_df.show()
```
