```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('example').getOrCreate()
```

```python
from pyspark.sql.functions import sum
```

```python
# Sample data
data = [
    ("John", "CA", 100),
    ("Alice", "NY", 200),
    ("John", "CA", 300),
    ("Alice", "NY", 150),
    ("Mark", "TX", 400),
]
```

```python
columns = ['customer', 'state', 'amount']
```

```python
df = spark.createDataFrame(data, columns)
```

```python
df.show()
```

```python
# Question 1. Write PySpark code to get the total amount spent by each customer, and order the result in descending order of the total amount.

result = df.groupBy('customer').agg(sum('amount').alias('Total_amount')).sort('Total_amount', ascending=False)
```
